## Credit Card Fraud Detection

importing the dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import warnings
warnings.filterwarnings('ignore')
```

```python
Fraud=credits[credits['Class']==1]
legit=credits[credits['Class']==0]
```

## Balancing the data

Build a sample dataset containing similar distribution of normal and fraudulent transaction

```python
legit_sample=legit.sample(n=492)
```

```python
new_data=pd.concat([legit_sample,Fraud],axis=0)
```

```python
new_data.head()
```

```python
new_data.tail()
```

```python
new_data.groupby('Class').mean()
```

Splitting the data in features and target variable

```python
x=new_data.drop('Class',axis=1)
y=new_data['Class']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.20,stratify=y, random_state=42)
```

```python
model=LogisticRegression()
model.fit(X_train,y_train)
```

```python
Y_train_pred=model.predict(X_train)
train_acc=accuracy_score(Y_train_pred,y_train)
```

```python
train_acc
```

```python
Y_test_pred=model.predict(X_test)
test_acc=accuracy_score(Y_test_pred,y_test)
```

```python
test_acc
```

## Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier=GridSearchCV(model,
                        {
                            'penalty':['l1','l2','elasticnet','none'],
                            'C':[0.01,0.1,1,10,100],
                            'l1_ratio':[0,0.5,1]
                        },cv=5)
```

```python
classifier.fit(x,y)
```

```python
pd.DataFrame(classifier.cv_results_).sort_values('mean_test_score',ascending=False).head()
```

Model after Tuning

```python
Hyper_model=LogisticRegression(C=0.01,l1_ratio=0,penalty='l2')
```

```python
Hyper_model.fit(X_train,y_train)
```

```python
Hyper_pred=Hyper_model.predict(X_train)
Hyper_acc=accuracy_score(Hyper_pred,y_train)
Hyper_acc
```

```python
Hyper_Test_pred=Hyper_model.predict(X_test)
Hyper_Test_acc=accuracy_score(Hyper_Test_pred,y_test)
Hyper_Test_acc
```
